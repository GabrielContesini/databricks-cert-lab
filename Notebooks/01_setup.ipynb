```python
%sql
MERGE INTO cert.bronze.minha_primeira_tabela AS destino
USING (SELECT * FROM cert.bronze.minha_primeira_tabela LIMIT 1) AS origem
ON destino.categoria = origem.categoria
WHEN MATCHED THEN UPDATE SET *
WHEN NOT MATCHED THEN INSERT *;
```
